# 1. Rossmann-Daten vorbereiten

Ziel: Die Originaldatei auf Filialen 1, 2 und 3 reduzieren. Dieser optionale Schritt schreibt `data/rossmann/train_filtered.csv` neu. Die Datei ist bereits enthalten.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
if not (ROOT / "src").is_dir():
    raise RuntimeError("Notebook im Projektordner oder in notebooks/ starten.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
import pandas as pd

source = pd.read_csv(ROOT / "data/rossmann/train.csv", low_memory=False)
selected_stores = [1, 2, 3]
filtered = source.loc[source["Store"].isin(selected_stores)].copy()
output = ROOT / "data/rossmann/train_filtered.csv"
filtered.to_csv(output, index=False)
print(f"Original: {len(source):,} Zeilen → Auswahl: {len(filtered):,} Zeilen")
filtered.head()
```

`isin` prüft die Filialnummern. `low_memory=False` lässt pandas die Typen ohne stückweise Typinferenz bestimmen; es repariert keine fehlerhaften Werte. `index=False` verhindert eine zusätzliche CSV-Spalte für den DataFrame-Index.
