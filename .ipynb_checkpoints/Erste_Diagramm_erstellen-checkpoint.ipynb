```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df_telco = pd.read_csv("data/telco/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df_rossmann_filtered = pd.read_csv("data/rossmann/train_filtered.csv") 
```

```python
print("\n" + "="*40 +"\nAnalyse: Telco Costumer Churn\n"+"="*40)
print("\n--- Datentyp ---\n")
print(df_telco.dtypes)

print("\n--- Fehlende Werte (Absolute Anzahl) ---\n")
print(df_telco.isnull().sum()) # nur null oder nah
```

```python
print("\n" + "="*40 +"\nAnalyse: Rossmann Daten\n"+"="*40)
print("\n--- Datentyp ---\n")
print(df_rossmann_filtered.dtypes)
print("\n--- Fehlende Werte (Absolute Anzahl) ---\n")
print(df_rossmann_filtered.isnull().sum())
```

```python
plt.figure(figsize=(6, 4))
# hue='Churn' und legend=False hinzugefügt für modernere Seaborn-Versionen
sns.countplot(data=df_telco, x='Churn', hue='Churn', palette='Set2', legend=False) #Legend = gibt noch ein kleines box oben
plt.title('Telco: Wie viele Kunden haben gekündigt (Churn)?')
plt.xlabel('Gekündigt?')
plt.ylabel('Anzahl Kunden')
plt.tight_layout() #Sorgt dafür, dass nichts vom Text abgeschnitten wird
plt.savefig(r'C:\mein data science projekt\Churun.png', dpi=300, bbox_inches='tight')
plt.show() #bringt das fertige Bild auf deinen Bildschirm.
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df_telco, x='Contract', hue='Churn', palette='Set1')
plt.title('Telco: Kündigungen nach Vertragsart')
plt.xlabel('Vertragstyp')
plt.ylabel('Anzahl Kunden')
plt.tight_layout()
plt.savefig('Kündigungen nach Vertragsart.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df_rossmann_filtered, x='Sales', kde=True, bins=30, color='purple') #kde = kernek density Estimate (Mcht eine Linie über das Diagram)
plt.title('Rossmann: Verteilung der täglichen Umsätze (Stores 1-3)')
plt.xlabel('Umsatz in €')
plt.ylabel('Häufigkeit')
plt.tight_layout()
plt.savefig('Rossmann Verteilung der täglichen Umsätze (Stores 1-3).png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
df_rossmann_filtered['Date'] = pd.to_datetime(df_rossmann_filtered['Date'])

plt.figure(figsize=(14, 5))
# .astype(str) sorgt dafür, dass Store 1, 2 und 3 komplett eigene Farben bekommen
sns.lineplot(data=df_rossmann_filtered, x='Date', y='Sales', hue=df_rossmann_filtered['Store'].astype(str), palette='tab10')
plt.title('Rossmann: Umsatzverlauf der Stores 1, 2 und 3')
plt.xlabel('Datum')
plt.ylabel('Umsatz in €')
plt.legend(title='Store ID')
plt.tight_layout()
plt.savefig('Rossmann Umsatzverlauf der Stores 1, 2 und 3.png', dpi=300, bbox_inches='tight')
plt.show()
# Countpilot kein y weil computer hohe der Balken selbst zählt 
# linepilot x und y und  hue geschäfte zu trännen 
```
