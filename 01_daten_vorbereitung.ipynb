```python
import pandas as pd

# 1. Die große Rossmann-Datei laden
print("Lade die Rossmann-Datei... (dauert 2-3 Sekunden)")
df_rossmann = pd.read_csv("data/rossmann/train.csv", low_memory=False) #unterschiedliche Datentyp in eine Spalte (Low_memory) vermeidet 
#fehler und bearbeitet besser und nimmt auch mehr Platz in Featplatt

# 2. Filtern auf die Stores 1, 2 und 3 (wie in deinen Zielen gewünscht)
# Du kannst die Zahlen hier beliebig ändern!
stores_auswahl = [1, 2, 3]
df_filtered = df_rossmann[df_rossmann['Store'].isin(stores_auswahl)]

# 3. Den gefilterten, kleinen Datensatz neu abspeichern
output_path = "data/rossmann/train_filtered.csv"
df_filtered.to_csv(output_path, index=False) # index = panda baut tabelle mit zusätzlicher zeilennummer.

print(f"Erfolg! Daten gefiltert. Ursprüngliche Zeilen: {len(df_rossmann)} -> Neue Zeilen: {len(df_filtered)}")
print(f"Die gefilterte Datei wurde unter '{output_path}' gespeichert.")
```
